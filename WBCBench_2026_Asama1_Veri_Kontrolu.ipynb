# WBCBench 2026 — Aşama 1: Veri Kontrolü

Bu notebook yalnızca veri klasörlerinin ve train/val/test dağılımının doğru olduğunu kontrol eder.

**Bu aşamada model eğitimi ve GPU kullanımı yoktur.**

Plan: Veri kontrolü → ConvNeXt-Tiny → ResNet-50 → RETFound Linear Probe → RETFound Partial Fine-Tuning.

```python
from pathlib import Path

# WBCBench processed dataset
DATA_DIR = Path("wbc_project/data/processed")

print("Data directory:", DATA_DIR.resolve())
print("Exists:", DATA_DIR.exists())
```

```python
IMAGE_EXTENSIONS = {'.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff'}

for split in ["train", "val", "test"]:
    split_dir = DATA_DIR / split
    print(f"\n--- {split.upper()} ---")

    if not split_dir.exists():
        print("NOT FOUND:", split_dir)
        continue

    classes = sorted([p for p in split_dir.iterdir() if p.is_dir()])
    total = 0

    for class_dir in classes:
        images = [
            p for p in class_dir.iterdir()
            if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
        ]
        print(f"{class_dir.name:20s}: {len(images)}")
        total += len(images)

    print("TOTAL:", total)
```
